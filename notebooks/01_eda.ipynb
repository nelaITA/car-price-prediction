```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["priceUSD"], bins=50, ax=ax[0])
sns.histplot(np.log1p(df["priceUSD"]), bins=50, ax=ax[1])
ax[0].set_title("priceUSD"); ax[1].set_title("log(priceUSD)")
plt.show()
```

Cijena je jako nakošena udesno, većina automobila je jeftina, a mali broj je vrlo skup.
